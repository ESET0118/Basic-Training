```python
import numpy as np
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('data.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df=df.dropna()
df.head()
```

```python
df.Make.value_counts().nlargest(40).plot(kind='bar',figsize=(10,5))
plt.title('Number of Cars by Make')
plt.ylabel('Number of Cars')
plt.xlabel('Car Make')
```

```python
df.head()
```

```python
df['HP*Cylinders']=df['Engine HP']*df['Engine Cylinders']
df.head()
```

```python
def mileageCategory(mileage):
    if mileage < 20:
        return 'Low'
    elif 20 <= mileage < 30:
        return 'Medium'
    else:
        return 'High'
    
df['Mileage Category']=df['highway MPG'].apply(mileageCategory)
df.head()
```

```python
df.groupby('Mileage Category').size()
```

```python
df.groupby('Transmission Type').agg({'Engine HP':'mean','Engine Cylinders':'mean','highway MPG':'mean','city mpg':'mean'})
```
