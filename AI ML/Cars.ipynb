```python
import numpy as np
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('data.csv')
df.head()
df.shape
df.info()
```

```python
df.isnull().sum()
```

```python
df=df.dropna()
df.head()
```

```python
df.head()
```

```python
df['HP*Cylinders']=df['Engine HP']*df['Engine Cylinders']
df.head()
```

```python
def mileageCategory(mileage):
    if mileage < 20:
        return 'Low'
    elif 20 <= mileage < 30:
        return 'Medium'
    else:
        return 'High'
    
df['Mileage Category']=df['highway MPG'].apply(mileageCategory)
df.head()
```

```python
df.groupby('Mileage Category').size()
```

```python
df.groupby('Transmission Type').agg({'Engine HP':'mean','Engine Cylinders':'mean','highway MPG':'mean','city mpg':'mean'})
```

```python
df = df.drop(['Engine Fuel Type', 'Market Category', 'Vehicle Size', 'Vehicle Style', 'Popularity', 'Number of Doors'], axis=1)
df.head()
```

```python
df.shape
```

```python
df=df.rename(columns={'Engine HP':'HP','Engine Cylinders':'Cylinders','MSRP':'Price'})
df.head()
```

```python
duplicate=[df.duplicated()]
df.drop_duplicates(inplace=True)
df.shape
```
