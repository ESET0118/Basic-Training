```python
%pip install scikit-learn

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import sklearn as sk
import sklearn.linear_model as lm
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
df=pd.read_csv('Salary_dataset.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.info()
df.head()
```

```python
x=df['YearsExperience']
y=df['Salary']
x.head()
```

```python
y.head()
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
model=lm.LinearRegression()

x_train=x_train.values.reshape(-1,1)
x_test=x_test.values.reshape(-1,1)
y_train=y_train.values.reshape(-1,1)
y_test=y_test.values.reshape(-1,1)

x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
model.fit(x_train,y_train)
y_predict=model.predict(x_test)
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_predict, alpha=0.7, label='Actual vs. Predicted')
min_val = min(y_test.min(), y_predict.min())
max_val = max(y_test.max(), y_predict.max())
plt.plot([min_val, max_val], [min_val, max_val], color='red', linestyle='--', label='Perfect Prediction (y_test = y_predict)')
plt.xlabel("True Values (y_test)")
plt.ylabel("Predicted Values (y_predict)")
plt.title("True vs. Predicted Values with Perfect Prediction Line")
plt.legend()
plt.grid(True)
plt.show()
```

```python


r2score = r2_score(y_test, y_predict)
print("R2 Score:", r2score)
```

```python
mean_squared_error = sk.metrics.mean_squared_error(y_test, y_predict)
mean_squared_error
```
